Step 1: Define Floyd-Warshall Algorithm with Path Reconstruction and Negative Cycle Detection

```python
def floyd_warshall(graph, V):
    dist = [[float('inf')] * V for _ in range(V)]
    next_node = [[-1] * V for _ in range(V)]

    for u in range(V):
        dist[u][u] = 0

    for u in graph:
        for v, w in graph[u]:
            dist[u][v] = w
            next_node[u][v] = v

    for k in range(V):
        for i in range(V):
            for j in range(V):
                if dist[i][k] + dist[k][j] < dist[i][j]:
                    dist[i][j] = dist[i][k] + dist[k][j]
                    next_node[i][j] = next_node[i][k]

    for i in range(V):
        if dist[i][i] < 0:
            print("Negative cycle detected")
            return dist, next_node, True

    print("No negative cycle detected.")
    return dist, next_node, False
```

Step 2: Define Path Reconstruction Function

```python
def reconstruct_path(u, v, next_node):
    if next_node[u][v] == -1:
        return []
    path = [u]
    while u != v:
        u = next_node[u][v]
        path.append(u)
    return path
```

Step 3: Sample Graph and Execute the Algorithm

```python
graph = {
    0: [(1, 2), (3, 9)],
    1: [(2, -5)],
    2: [(3, 2)],
    3: [(1, 1)]
}

V = 4
dist, next_node, has_cycle = floyd_warshall(graph, V)

print("\nDistance Matrix:")
for row in dist:
    print(row)

print("\nPath from 0 to 3:", reconstruct_path(0, 3, next_node))
```
