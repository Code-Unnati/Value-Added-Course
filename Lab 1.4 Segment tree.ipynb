### Objective
Design and implement a segment tree to support range minimum queries and point updates. Test and evaluate the implementation on dynamic data streams.



#### Equipment Required
- Python (Jupyter Notebook)
- No external libraries required



#### Prerequisites
- Understanding of arrays and range queries
- Basic programming knowledge in Python
- Familiarity with recursive functions



#### Problem Statement
Develop a segment tree that allows:
1. Efficient range minimum queries (RMQ)
2. Point updates on dynamic data
Then, test the segment tree with a stream of data changes and query results.


```python

#### Step 1: Segment Tree Class Definition

class SegmentTree:
    def __init__(self, data):
        self.n = len(data)
        self.tree = [float('inf')] * (4 * self.n)
        self.build(data, 0, 0, self.n - 1)

    def build(self, data, node, left, right):
        if left == right:
            self.tree[node] = data[left]
        else:
            mid = (left + right) // 2
            self.build(data, 2 * node + 1, left, mid)
            self.build(data, 2 * node + 2, mid + 1, right)
            self.tree[node] = min(self.tree[2 * node + 1], self.tree[2 * node + 2])

    def range_min(self, l, r):
        return self._range_min(0, 0, self.n - 1, l, r)

    def _range_min(self, node, left, right, l, r):
        if l > right or r < left:
            return float('inf')
        if l <= left and right <= r:
            return self.tree[node]
        mid = (left + right) // 2
        return min(
            self._range_min(2 * node + 1, left, mid, l, r),
            self._range_min(2 * node + 2, mid + 1, right, l, r)
        )

    def update(self, index, value):
        self._update(0, 0, self.n - 1, index, value)

    def _update(self, node, left, right, index, value):
        if left == right:
            self.tree[node] = value
        else:
            mid = (left + right) // 2
            if index <= mid:
                self._update(2 * node + 1, left, mid, index, value)
            else:
                self._update(2 * node + 2, mid + 1, right, index, value)
            self.tree[node] = min(self.tree[2 * node + 1], self.tree[2 * node + 2])
```

```python
#### Step 2: Testing the Segment Tree

data_stream = [5, 2, 6, 3, 1, 7, 4]
tree = SegmentTree(data_stream)

print("Initial Range Minimum (0-6):", tree.range_min(0, 6))
print("Range Minimum (2-4):", tree.range_min(2, 4))

# Perform point update
print("\nUpdating index 3 to 0")
tree.update(3, 0)
print("Updated Range Minimum (2-4):", tree.range_min(2, 4))

print("\nUpdating index 4 to 8")
tree.update(4, 8)
print("Updated Range Minimum (0-6):", tree.range_min(0, 6))
```
