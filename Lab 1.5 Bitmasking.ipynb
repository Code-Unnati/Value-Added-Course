### Objective
Implement a scheduling solution using bitmask dynamic programming (DP) to assign workers to tasks efficiently.



#### Equipment Required
- Python (Jupyter Notebook)
- No external libraries required



#### Prerequisites
- Basic understanding of dynamic programming
- Familiarity with bit manipulation
- Knowledge of recursive functions and memoization



#### Problem Statement
Given a matrix `cost[i][j]` where `cost[i][j]` represents the cost of assigning worker `i` to task `j`, implement a bitmask DP solution to find the minimum cost to assign all tasks to workers such that each worker is assigned exactly one task and each task is assigned to one worker.


```python

#### Step 1: Define the Bitmask DP Function

import sys

def min_cost_assignment(cost):
    n = len(cost)
    memo = [-1] * (1 << n)

    def dp(mask):
        i = bin(mask).count('1')
        if i == n:
            return 0
        if memo[mask] != -1:
            return memo[mask]

        ans = sys.maxsize
        for j in range(n):
            if not (mask & (1 << j)):
                ans = min(ans, cost[i][j] + dp(mask | (1 << j)))
        memo[mask] = ans
        return ans

    return dp(0)
```

```python
#### Step 2: Define a Sample Cost Matrix and Run the Function

# Example cost matrix
cost_matrix = [
    [9, 2, 7, 8],
    [6, 4, 3, 7],
    [5, 8, 1, 8],
    [7, 6, 9, 4]
]

min_cost = min_cost_assignment(cost_matrix)
print("Minimum Cost to Assign Tasks:", min_cost)
```
