Implement Bellman-Ford algorithms

Step 1: Define the Bellman-Ford Algorithm

```python
def bellman_ford(graph, start, V):
    distance = [float('inf')] * V
    distance[start] = 0

    for _ in range(V - 1):
        for u in graph:
            for v, w in graph[u]:
                if distance[u] != float('inf') and distance[u] + w < distance[v]:
                    distance[v] = distance[u] + w

    for u in graph:
        for v, w in graph[u]:
            if distance[u] != float('inf') and distance[u] + w < distance[v]:
                print("Graph contains a negative weight cycle")
                return None

    return distance
```

Step 2: Create a Sample Graph with Negative Weights

```python
graph = {
    0: [(1, 6), (2, 7)],
    1: [(2, 8), (3, 5), (4, -4)],
    2: [(3, -3), (4, 9)],
    3: [(1, -2)],
    4: [(0, 2), (3, 7)]
}
V = 5
```

Step 3: Execute the Bellman-Ford Algorithm and Display Output

```python
start_node = 0
output = bellman_ford(graph, start_node, V)

if output:
    print(f"\nShortest distances from node {start_node}:")
    for i, dist in enumerate(output):
        print(f"Node {i}: {dist}")
```
