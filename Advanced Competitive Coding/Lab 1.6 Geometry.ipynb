### Objective
Write a function to detect whether two line segments intersect and test the algorithm against various edge cases such as collinear and overlapping segments.



#### Equipment Required
- Python (Jupyter Notebook)
- No external libraries required



#### Prerequisites
- Understanding of basic geometry concepts
- Familiarity with programming conditional logic



#### Problem Statement
Given two line segments, each defined by two endpoints, determine whether the segments intersect. The solution must handle edge cases such as overlapping and collinear segments.


```python

#### Step 1: Define Orientation and On-Segment Functions

# Helper function to calculate orientation
# 0 --> Collinear, 1 --> Clockwise, 2 --> Counterclockwise
def orientation(p, q, r):
    val = (q[1] - p[1]) * (r[0] - q[0]) - (q[0] - p[0]) * (r[1] - q[1])
    if val == 0:
        return 0
    return 1 if val > 0 else 2

# Helper function to check if point q lies on segment pr
def on_segment(p, q, r):
    if min(p[0], r[0]) <= q[0] <= max(p[0], r[0]) and min(p[1], r[1]) <= q[1] <= max(p[1], r[1]):
        return True
    return False
```

```python
#### Step 2: Define the Segment Intersection Function

def do_intersect(p1, q1, p2, q2):
    o1 = orientation(p1, q1, p2)
    o2 = orientation(p1, q1, q2)
    o3 = orientation(p2, q2, p1)
    o4 = orientation(p2, q2, q1)

    # General case
    if o1 != o2 and o3 != o4:
        return True

    # Special Cases
    if o1 == 0 and on_segment(p1, p2, q1): return True
    if o2 == 0 and on_segment(p1, q2, q1): return True
    if o3 == 0 and on_segment(p2, p1, q2): return True
    if o4 == 0 and on_segment(p2, q1, q2): return True

    return False
```

```python
#### Step 3: Test the Function with Edge Cases

# Test 1: Intersecting segments
print(do_intersect((1, 1), (10, 1), (1, 2), (10, 2)))  # False
print(do_intersect((10, 0), (0, 10), (0, 0), (10, 10)))  # True

# Test 2: Collinear overlapping segments
print(do_intersect((1, 1), (5, 5), (2, 2), (6, 6)))  # True

# Test 3: Collinear non-overlapping segments
print(do_intersect((1, 1), (5, 5), (6, 6), (8, 8)))  # False
```
